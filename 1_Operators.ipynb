## **Python Operators**

```python
# Arithmetic operators
# +, -, *, / ,** ,//,%

print(10 + 3) 
print(10 - 3) 
print(10 * 3) 
print(10 / 3)
print(10 ** 3) # Pwo = 10*10*10 = 1000
print(10 // 3) # Q = 3
print(10 % 3) # R = 1
```

```python
# Assignment operators
# =, +=, -=, /=, *= , **=, //=, %=

a = 10
print(a)

a += 5
print(a)

a -= 4
print(a)

a *= 5
print(a)

a /= 6
print(a)

a **= 7
print(a)

a //= 8
print(a)

a %= 9
print(a)
```

```python
# Comparison operators  (True or False)
# ==, !=, >, <, <=, >=

print(10 == 3) 
print(10 != 3) 
print(10 > 3) 
print(10 < 3)
print(10 >= 3)
print(10 <= 3) 
```

```python
# Logical operators   (True or False)
# and, or, not


print("-----AND-----")

print(10 == 10 and 6 > 4)# T and T = T
print(10 != 10 and 6 > 4)# F and T = F
print(10 == 10 and 6 < 4)# T and F = F
print(10 != 10 and 6 < 4)# F and F =F


print("-----OR-----")

print(10 == 10 or 6 > 4)# T or T = T
print(10 != 10 or 6 > 4)# F or T = T
print(10 == 10 or 6 < 4)# T or F = T
print(10 != 10 or 6 < 4)# F or F = F


print("-----NOT-----")

print(not 10 == 10) # F
print(not 10 != 10) # T
```
