## **If Elif Else**  (Decision statement)  

### Syntax  

* if \<condition>:  
* elif \<condition>:  
* else:  

```python
a =10

if a>100:
    print(a,"is bigger than a")

print("This line is not work")
```

```python
a = 70

if a>100:
    print(a,"is bigger than a")

else:
    print(a,"is smaller than a")

print("This line is not work")
```

```python
a = 70

if a>100:
    print(a,"is bigger than a")

elif a==70:
    print(a,"is equal to a")

else:
    print(a,"is smaller than a")

print("This line is not work")
```

### **If ,elif, else rules.**  

* if (Only once per code block(Mandtory))  
* elif(as many time as per code block(optional))  
* else(Only once per code black(optional))
