## **Task using for loop**

```python
row=int(input("enter the row: "))
symbol=input("enter the symbol: ")

for x in range(row):
    for y in range(row):
         print(symbol,end="")
    print()
print("loop is over")
```

## **Task using while loop**

```python
x=int(input("Enter the row: "))
symbol=input("Give any symbol: ")

a=1
while a<=x:
    b=1
    while b<=a:
        print(symbol,end="")
        b +=1
    print()
    a +=1
print("Loop is over")
```
