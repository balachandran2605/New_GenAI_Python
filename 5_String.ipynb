### **isalpha()** 

```python
a="helloworld"
x=a.isalpha()
print(x)

print("hellobala".isalpha())
```

### **isdigit()**

```python
a="1236654"
x=a.isdigit()
print(x)
```

### **isalnum()**

```python
a="bala2605"
x=a.isalnum()
print(x)

print("bala".isalnum())
print("123".isalnum())
```

### **islower()**

```python
a="hellobala"
x=a.islower()
print(x)

print("Hello".islower())
```

### **isupper()**

```python
a="HELLOBALA"
x=a.isupper()
print(x)

print("HEllobalA".isupper())
```

### **lower()**

```python
a="HELLOSANJU"
x=a.lower()
print(x)

print("hellosanju".lower())
```

### **upper()**

```python
a="hellokarthi"
x=a.upper()
print(x)
```

### **title()**

```python
a="leave letter"
x=a.title()
print(x)
```

### **swapcase()**

```python
a="HelLO My NAme is ChANdran"
x=a.swapcase()
print(x)
```

###  **Split()**

```python
a="hi how are you i am fine da"
x=a.split("a")
print(x)
```

```python
a="onomatopoeiao"
x=a.split('o')
print(x)
```

### **join()**

```python
a=["How","are","you","Bala","I am","fine","da"]
x=" ".join(a)
print(x)
```

```python
# Interview question

text="hi da how are you"
words=text.split()
reverse=words[::-1]
result=" ".join(reverse)
print(result)
```

### **replace()**

```python
a="Hi da My name is bala!I am from chennai,I am currently work with Gen AI developer"
x=a.replace('a','--')
print(x)
```

### **count()**

```python
a="Hi da My name is bala!I am from chennai,I am currently work with Gen AI developer"
x=a.count('r')
print(x)
```

### **find()**

```python
a="Cheif Minister of Tamilnadu Thiru.C.Joshep Vijay"
x=a.find('y')
print(x)
```

### **ends with()**

```python
a="Hello World"
x=a.endswith('ld')
print(x)

y=a.endswith('gh')
print(y)
```

```python
mylist=["bala.py","dinesh.txt","karthi.py","mala.txt"]
for each in mylist:
    if each.endswith(".py"):
        print(each)
    
```

### **Startswith()**

```python
a="Hello Bala"
x=a.startswith('H')
print(x)

y=a.startswith('g')
print(y)
```

```python
mylist=["bala.py","dinesh.txt","karthi.py","mala.txt"]
for each in mylist:
    if each.startswith("b"):
        print(each)
```
