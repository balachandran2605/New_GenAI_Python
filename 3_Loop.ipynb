## **Python Loop**  

* *While loopcondition statements*  
* *Syntax*
* *while \<condition>:*  

```python
a=1

while a<=10:
    print(a,end=" ")
    a += 1
```

```python
a=60

while a>=50:
    print(a,end=" ")
    a -=1
print("\n Loop Over")
```

```python
a=100

while a>=0:
    print(a,end=" ")
    a -=10
print("\n Loop Over")
```

## **break** – *stop the lop*

```python
a=1

while a>=0:
    if a==20:
        print(a)
        break
    print(a,end=" ")
    a +=1
print("\n Loop Over")
```

## **continue** – *go back to the top of my loop*

```python
a=1

while a<20:
    if a % 3==0:
        print(a,end=" ")
        a +=1
        continue
    a +=1
print("\n Loop over")
  
```

## **Nested While Loop**

```python
"""
12345
12345
12345
12345
12345

"""

a=1

while a<=5:
    print(a,end='')
    a +=1
```

```python
a=1

while a<=5:
    b=1
    while b<=5:
        print(b,end="")
        b +=1
    print()
    a +=1

print("LOOP OVER")
```

## **For loop**  
**range()**–is a python standard function used for numbers  

*Syntax*  
### **for \<variable_Name> in \<str\List\Tuple\Set\Dict>:**

```python
for x in "Hello World":
    print(x,end="")
```

```python
mylist =["hello",100,True,30.55,"bala"]

for x in mylist:
    print(x,end=",")
```

## **range()**–is a python standard function used for numbers

```python
for  x in range(20+1):
    print(x,end=" ")
```

```python
for x in range(1,30):
    print(x,end=" ")
```

```python
for x in range(10,100+1,10):
    print(x,end=" ")
```

```python
for x in range(100,0,-20):
    print(x,end=" ")
```

## **for with else**  

* If Loop completes successfully, else will work  
* If loop is unsuccessfully , else will not work  

```python
for x in "hello world":
    print(x,end="")

else:
    print("\n loop is successfully completed")
```

```python
for x in "helloworld":
    if x == " ":
     break
    print(x,end="")

else:
    print("\n loop is over")
```

```python
mylist=[2,6,9,10,15,11,20,16,97]

for each in mylist:
    if each % 5 ==0:
     print(each," divisible by 5")

else:
    print("each is not divisible by 5")
                                                        
```

## **Nested Loop**

```python
"""
12345
12345
12345
12345
12345
12345
"""

for x in range(5):
    for y in range(1,6+1):
        print(y,end="")
    print()
```

```python
row=int(input("Enter the row: "))
column=int(input("Enter the column: "))

for x in range(row):
    for y in range(1,column+1):
        print(y,end="")
    print()
```
