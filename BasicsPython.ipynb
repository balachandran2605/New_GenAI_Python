# **Python Data Types**
* Integer – 1, 1244, -1234  
* Float – 1.0, 1e10, 123.324  
* complex – 1+0j (j represents that its complex)  
* string – “Hello world”, “a”, ‘a’  
* boolean – True, False  

```python
# Integer – 1, 1244, -1234 

a = 100
print(a,type(a))

a1 = -200
print(a1,type(a1))
```

```python
# Float – 1.0, 1e10, 123.324  

b = 30.0
print(b,type(b))

b1 = 2e7
print(b1,type(b1))

b2 = -23.9
print(b2,type(b2))


```

```python
# string – “Hello world”, “a”, ‘a’  

c = "Hello world"
print(c,type(c))

c1 = "500+33"
print(c1,type(c1))
```

```python
# complex – 1+0j (j represents that its complex)  

d = 20j
print(d,type(d))
```

```python
#  boolean – True, False 

e = True
print(e,type(e))

e1 = False
print(e1,type(e1))
```

### **Escape sequences**


</center>

* \ - for newline in editor  
* \\\ - prints \  
* \’ – prints ‘  
* \” – prints “  
* \n – prints a new line  
* \t – prints a tab  

```python
# \\ - prints \ 
print("Hello \\world\\")
print("100\\2oo")

# \' - prints'
print('I\'am Bala')

# \"-prints"
print("\"Hello\" \"world\"")

# \n - prints a new line
print("Hello\nBala")

# \t - printa new tab
print("Hell\tBala")
```
